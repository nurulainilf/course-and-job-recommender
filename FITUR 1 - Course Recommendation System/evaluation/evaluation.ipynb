```python
import pandas as pd
```

```python
df = pd.read_csv('ground_truth.csv')
df.head(5)
```

```python
grouped = df.groupby(['Skills', 'Job Industry'])

records = []

for (skills, industry), group in grouped:
    top_10 = group.sort_values(by='Similarity Score', ascending=False).head(10)
    relevant = top_10['Ground Truth'].sum()
    precision = relevant / 10

    records.append({
        'Skills': skills,
        'Job Industry': industry,
        'Precision@10': precision
    })

result_df = pd.DataFrame(records)

test_case_fields = [
    "Project Management",
    "Front-end Development",
    "Cybersecurity",
    "Digital Marketing",
    "Data Science",
    "UI/UX Design",
    "Game Development"
]

result_df['Test Case Field'] = test_case_fields
result_df = result_df[['Test Case Field', 'Precision@10']]
result_df
```

```python
average_precision = result_df['Precision@10'].mean()
print(f"\nAverage Precision@10: {average_precision:.4f}")
```
